```r
finish_distribution <- march_data_tidy |>
    filter(!is.na(tourney_finish)) |>
    group_by(tourney_finish) |>
    summarize(count = n()) |>
    mutate(proportion = count / sum(count))

ggplot(finish_distribution, aes(x = tourney_finish, y = proportion)) +
    geom_bar(stat = "identity") +
    labs(title = "Proportion of Teams by Tournament Finish", 
         x = "Tournament Finish", 
         y = "proportion of Teams") +
    theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

Visualization 1: This bar plot compares the average tournament seed across different tournament finish categories. It helps us examine whether teams with lower seed numbers tend to advance further in the tournament. This is relevant to our research question because tournament seed is one of the predictors. 

```r
seed_finish_summary <- march_data_tidy |>
    filter(!is.na(tourney_finish)) |>
    group_by(tourney_finish) |>
    summarize(mean_seed = mean(seed, na.rm = TRUE))

ggplot(seed_finish_summary, aes(x = tourney_finish, y = mean_seed)) +
    geom_bar(stat = "identity") +
    labs(title = "Average Tournament Seed by Tournament Finish", 
         x = "Tournament Finish", 
         y = "Average Tournament Seed") +
    theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

Visualization 2: This bar plot compares the average conference win percentage across different tournament finish categories. It helps us examine whether teams with higher conference win percentages tend to achieve better tournament results. This is relevant to our research question because conference win percentage is the second predictor. 

```r
conf_finish_summary <- march_data_tidy |>
    filter(!is.na(tourney_finish)) |>
    group_by(tourney_finish) |>
    summarize(mean_conf_wins_pct = mean(conf_wins_pct, na.rm = TRUE))

ggplot(conf_finish_summary, aes(x = tourney_finish, y = mean_conf_wins_pct)) +
    geom_bar(stat = "identity") +
    labs(title = "Average Conference Win Percentage by Tournament Finish", 
         x = "Tournament Finish", 
         y = "Average Conference Win Percentage") +
    theme(axis.text.x = element_text(angle = 45, hjust = 1))
```
